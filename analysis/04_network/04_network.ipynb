## Setting

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = (Path.cwd().parent).parent        

NETWORK_DIR = ROOT / "04_network"
DATA_DIR    = ROOT / "data"/ "04_network"

sys.path.insert(0, str(NETWORK_DIR))

from PCMCI import load_pcmci_dataframe, run_pcmci_gpdc, results_to_long_df

FIG_DIR = ROOT / "figures" / "04_network"
RES_DIR = ROOT / "results" / "04_network"

FIG_DIR.mkdir(parents=True, exist_ok=True)
RES_DIR.mkdir(parents=True, exist_ok=True)
```

## Action Binding

```python
# Cell 2  Action binding
csv_path = DATA_DIR / "action_binding_for_network.csv"
dataframe, regions, data = load_pcmci_dataframe(csv_path)

results = run_pcmci_gpdc(
    dataframe=dataframe,
    regions=regions,
    label="Action binding",
    tau_max=3,
    pc_alpha=0.05,
    alpha_level=0.05,
    save_fig_path=FIG_DIR / "action_binding.png",
    show_fig=True,
)

df_long = results_to_long_df(results, regions, tau_min=1, tau_max=3)
df_long.to_csv(RES_DIR / "action_binding.csv", index=False)
df_long.head(10)
```

## Outcome Binding

```python
# Cell 3  Outcome binding
csv_path = DATA_DIR / "outcome_binding_for_network.csv"
dataframe, regions, data = load_pcmci_dataframe(csv_path)

results = run_pcmci_gpdc(
    dataframe=dataframe,
    regions=regions,
    label="Outcome binding",
    tau_max=3,
    pc_alpha=0.1,
    alpha_level=0.1,
    save_fig_path=FIG_DIR / "outcome_binding.png",
    show_fig=True,
)

df_long = results_to_long_df(results, regions, tau_min=1, tau_max=3)
df_long.to_csv(RES_DIR / "outcome_binding.csv", index=False)
df_long.head(10)
```
