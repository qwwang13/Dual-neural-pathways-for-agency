## Input

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = (Path.cwd().parent).parent   
LMM_DIR = ROOT / "06_ieeg"  

sys.path.insert(0, str(LMM_DIR))

from ITPC_ieeg import export_angles_itpc_ieeg


DATA_DIR = ROOT / "data" / "06_ieeg" 
OUT_DIR  = ROOT / "data" / "06_ieeg"  / "plot_data_for_itpc"

FREQS = list(range(1, 80))
```

## Outcome binding

```python
outcome_jobs = [
    ("ieeg_PostcentralL.csv", (0.20, 0.30), []),
    ("ieeg_MTGR.csv",         (0.77, 0.85), []),
]

outcome_records = []
for fname, tw, excl in outcome_jobs:
    csv_path = DATA_DIR / fname
    angles_csv, itpc_csv, angles_df, itpc_mean_df = export_angles_itpc_ieeg(
        csv_path=csv_path,
        time_window=tw,
        out_dir=OUT_DIR,
        freqs=FREQS,
        exclude_subjects=excl,
    )
    outcome_records.append(
        {"file": fname, "time_window": tw, "angles_csv": str(angles_csv.relative_to(ROOT)), "itpc_csv": str(itpc_csv.relative_to(ROOT))}
    )

pd.DataFrame(outcome_records)
```
