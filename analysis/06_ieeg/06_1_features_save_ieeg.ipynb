## Input

```python
from pathlib import Path
import sys
import pandas as pd

ROOT = (Path.cwd().parent).parent
LMM_DIR = ROOT / "06_ieeg"
sys.path.insert(0, str(LMM_DIR))

from TF_features_ieeg import export_trial_features_ieeg

DATA_DIR = ROOT / "data" / "06_ieeg" 
OUT_DIR  = ROOT / "data" / "06_ieeg" / "feature"

FREQS = list(range(1, 80))
```

## Outcome binding

```python
outcome_jobs = [
    ("ieeg_MTGR.csv",         (0.77, 0.85)),
]

outcome_records = []
for fname, tw in outcome_jobs:
    out_csv, merged = export_trial_features_ieeg(
        csv_path=DATA_DIR / fname,
        time_window=tw,
        out_dir=OUT_DIR,
        freqs=FREQS,
    )
    outcome_records.append({"file": fname, "time_window": tw, "out_csv": str(out_csv.relative_to(ROOT))})

pd.DataFrame(outcome_records)
```
